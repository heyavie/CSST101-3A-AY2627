**10. Creating a Simple Rule-Based System**

```r
check_risk <- function(attendance, grade) {
  if (attendance < 75 & grade < 75) {
    return("High Risk")
  } else {
    return("Low Risk")
  }
}
```

**11. Apply the Rule to Multiple Students**

```r
student_risk <- function(attendance, grade) {
  if (attendance < 75 & grade < 75) {
    return("High Risk")
  }
  else if (attendance < 85 || grade < 80) {
    return("Moderate Risk")
  }
  else {
    return("Low Risk")
  }
}

students <- data.frame(
  Student_ID = c("S001", "S002", "S003", "S004", "S005"),
  Attendance = c(95, 72, 88, 79, 68),
  Grade = c(90, 70, 85, 77, 65)
)


students$Risk <- mapply(
  student_risk,
  students$Attendance,
  students$Grade
)

students
```

**14. Add a Recommendation**

```r
get_recommendation <- function(risk_level) {
  if (risk_level == "High Risk") {
    return("Immediate academic intervention")
  } else if (risk_level == "Moderate Risk") {
    return("Academic monitoring")
  } else {
    return("Continue regular monitoring")

  }
}

students$Recommendation <- sapply(students$Risk, get_recommendation)

students
```

**15. Final System**

```r
evaluate_student <- function(student_id, attendance, grade) {
  if (attendance < 75 || grade < 75) {
    risk_level <- "High Risk"
    recommendation <- "Immediate academic intervention"
  } else if (attendance < 85 || grade < 85) {
    risk_level <- "Moderate Risk"
    recommendation <- "Monitor progress and offer support"
  } else {
    risk_level <- "Low Risk"
    recommendation <- "On track"
  }

  cat("Student ID:", student_id, "\n")
  cat("Attendance:", paste0(attendance, "%"), "\n")
  cat("Grade:", grade, "\n\n")
  cat("Risk Level:", risk_level, "\n\n")
  cat("Recommendation:\n")
  cat(recommendation, "\n")
}

evaluate_student(student_id = "S002", attendance = 72, grade = 70)
```
