3. Numerical Operators

```r
x <- 10
y <- 3

add <- x + y
sub <- x - y
mult <- x * y
div <- x / y
mod <- x %% y
exp <- x ^ y

print(add)
print(sub)
print(mult)
print(div)
print(mod)
print(exp)
```
